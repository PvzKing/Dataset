# Editor isi `<think>` dataset own

Tulis sendiri isi tag `<think>` untuk setiap sampel. Setiap kali **Simpan**, teksnya otomatis masuk ke `data/train.jsonl` di repo GitHub.

**Sekali saja, sebelum mulai:**
1. Buat token GitHub di https://github.com/settings/personal-access-tokens/new: pilih *Only select repositories* → `PvzKing/Dataset`, lalu *Repository permissions* → **Contents: Read and write**.
2. Di Colab, klik ikon kunci 🔑 (**Secrets**) di panel kiri → *Add new secret* → nama `GITHUB_TOKEN`, isi dengan token tadi, lalu aktifkan *Notebook access*.

Lalu jalankan sel di bawah (runtime CPU biasa cukup). Setelah muncul link `https://....gradio.live`, buka link itu dan login dengan user `penulis` dan sandi yang kamu isi di `SANDI`.

```python
#@title ✍️ Buka editor <think>
NAMA = "Nama Anda"  #@param {type:"string"}
EMAIL = "email@contoh.com"  #@param {type:"string"}
SANDI = "ganti-sandi-ini"  #@param {type:"string"}
#@markdown Sandi ini melindungi link editor; siapa pun yang punya link dan sandi bisa menulis ke dataset.

import os, subprocess
from google.colab import userdata
try:
    os.environ["GITHUB_TOKEN"] = userdata.get("GITHUB_TOKEN")
except Exception:
    print("⚠️ Secret GITHUB_TOKEN belum ada atau belum diberi akses: perubahan tetap tersimpan di Colab, tapi tidak di-push.")
REPO = "/content/Dataset"
if os.path.isdir(REPO):
    subprocess.run(["git", "-C", REPO, "pull", "-q", "--autostash", "origin", "main"], check=False)
else:
    subprocess.run(["git", "clone", "-q", "https://github.com/PvzKing/Dataset.git", REPO], check=True)
subprocess.run(["git", "-C", REPO, "config", "user.name", NAMA], check=True)
subprocess.run(["git", "-C", REPO, "config", "user.email", EMAIL], check=True)
subprocess.run(["pip", "install", "-q", "gradio>=5,<7"], check=True)
assert SANDI and SANDI != "ganti-sandi-ini", "Isi SANDI dengan sandi buatanmu sendiri dulu."
%run /content/Dataset/scripts/editor_think.py --share --auth penulis:{SANDI}
```
